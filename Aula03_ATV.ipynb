```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**3 - x - 2

# Criando 50 pontos entre 1 e 2
x = np.linspace(1, 2, 50)
y = f(x)

plt.plot(x, y)
plt.axhline(0, color='black') # Linha do zero (eixo X)
plt.grid(True)
plt.title("Gráfico de f(x)")
plt.show()
```

# Desafio 1:
# Onde parece estar localizada a raiz?
Observando o gráfico, a linha cruza o eixo X (onde $f(x) = 0$) um pouco depois do 1.5. Portanto, a raiz parece estar localizada em aproximadamente $1.52$.
# Existe mais de uma raiz?
 No intervalo sugerido de $[1, 2]$, o gráfico mostra que a função é estritamente crescente, cruzando o eixo X apenas uma vez. Logo, existe apenas uma raiz real neste intervalo.

```python
a = 1
b = 2

fa = f(a)
fb = f(b)

print("f(1) =", fa)
print("f(2) =", fb)
print("Multiplicação:", fa * fb)

if fa * fb < 0:
    print("Tem mudança de sinal!")
else:
    print("Não tem mudança de sinal.")
```

# Desafio 2:
# Existe mudança de sinal?
Sim. Calculando os extremos, temos $f(1) = -2$ e $f(2) = 4$. Como $-2 \cdot 4 = -8$ (que é menor que 0), há uma mudança de sinal.
# O intervalo escolhido é adequado?
Sim, o intervalo $[1, 2]$ é perfeitamente adequado, pois a mudança de sinal garante (pelo Teorema do Valor Intermediário) que existe pelo menos uma raiz ali dentro.

```python
def f(x):
    return x**3 - x - 2

a = 1.0
b = 2.0
iteracoes = 10

# Cabeçalho da nossa tabela "manual"
print("Iter | a      | b      | c      | f(c)")
print("-" * 40)

for i in range(1, iteracoes + 1):
    c = (a + b) / 2
    fc = f(c)

    # Imprimindo os valores com 4 casas decimais para ficar alinhado
    print(f"{i:4} | {a:.4f} | {b:.4f} | {c:.4f} | {fc:.4f}")

    # Verificando para qual lado o intervalo vai fechar
    if f(a) * fc < 0:
        b = c
    else:
        a = c
```

# Desafios 3, 4 e 5:
# Quantas iterações foram necessárias para obter uma boa aproximação da raiz?
 Com 10 iterações, chegamos a um valor de $c$ muito próximo de $1.5205$, com $f(c)$ ficando bem pequeno (próximo de zero). Para aplicações comuns de engenharia, 10 a 15 iterações já costumam oferecer uma margem de erro aceitável para esse método.
# O intervalo diminuiu pela metade a cada iteração?
Sim. Essa é a mecânica fundamental do método da bissecção. O tamanho do intervalo inicial era 1 (de 1 a 2). Na iteração seguinte caiu para 0.5, depois 0.25, e assim sucessivamente.
# O resultado obtido parece coerente com o gráfico da função?
Totalmente coerente. A raiz encontrada numericamente (próxima a $1.52$) é exatamente o ponto onde visualizamos a curva cruzando o eixo horizontal no primeiro passo.
